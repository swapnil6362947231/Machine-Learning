```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy='median')
```

```python
df['Heart Rate (BPM)']=imputer.fit_transform(df[['Heart Rate (BPM)']]).ravel()
```

```python
df.info()
```

```python
df['Heart Rate (BPM)'].isnull().sum()
```

```python
imputer = SimpleImputer(strategy='mean')
```

```python
df['Blood Oxygen Level (%)']=imputer.fit_transform(df[['Blood Oxygen Level (%)']]).ravel()
```

```python
df['Blood Oxygen Level (%)'].isnull().sum()
```

```python
df['Activity Level']=SimpleImputer(strategy='most_frequent').fit_transform(df[['Activity Level']]).ravel()
```

```python
df['Activity Level'].isnull().sum()
```

```python
df['Stress Level']=SimpleImputer(strategy='constant', fill_value=5).fit_transform(df[['Stress Level']]).ravel()
```

```python
df['Stress Level'].isnull().sum()
```
